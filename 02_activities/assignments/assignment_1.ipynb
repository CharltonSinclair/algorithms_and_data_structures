## Problem Three: Move All Zeros to End
**Description**  
Given a list of integers, move all zeros to the end while maintaining the relative order of the non-zero elements. 

**Examples**
* Example 1:
    ```python
    Input: nums = [0, 1, 0, 3, 12]
    Output: [1, 3, 12, 0, 0]
    ```
* Example 2:
    ```python
    Input: nums = [4, 0, 5, 0, 0, 6]
    Output: [4, 5, 6, 0, 0, 0]
    ```

**Problem 3 Starter Code**

````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    """Move every 0 to the end of nums while keeping the non-zero values
    in their original order. The list is rearranged in place and returned."""
    write = 0  # index where the next non-zero value should be placed

    for read in range(len(nums)):  # read visits every index once, left to right
        if nums[read] != 0:
            # swap the non-zero value forward into the write position
            nums[write], nums[read] = nums[read], nums[write]
            write += 1  # the next non-zero value goes one slot further right

    return nums
````

````python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    """Move every 0 to the end of nums while keeping the non-zero values
    in their original order. The list is rearranged in place and returned."""
    write = 0  # index where the next non-zero value should be placed

    for read in range(len(nums)):  # read visits every index once, left to right
        if nums[read] != 0:
            # swap the non-zero value forward into the write position
            nums[write], nums[read] = nums[read], nums[write]
            write += 1  # the next non-zero value goes one slot further right

    return nums


# Check the solution on the provided examples (1-2) and my new examples (3-4)
examples = [
    ([0, 1, 0, 3, 12], [1, 3, 12, 0, 0]),          # Example 1 (provided)
    ([4, 0, 5, 0, 0, 6], [4, 5, 6, 0, 0, 0]),      # Example 2 (provided)
    ([0, 0, -3, 7, 0, -3], [-3, 7, -3, 0, 0, 0]),  # Example 3 (new)
    ([9, 2, 5], [9, 2, 5]),                        # Example 4 (new)
]

for nums, expected in examples:
    original = nums.copy()  # keep a copy, because the function changes nums in place
    result = move_zeros_to_end(nums)
    print(f"Input: {original} -> Output: {result}")
    assert result == expected, f"expected {expected}"

# Edge cases: an empty list, a single zero, and a list of only zeros
assert move_zeros_to_end([]) == []
assert move_zeros_to_end([0]) == [0]
assert move_zeros_to_end([0, 0, 0]) == [0, 0, 0]
print("All examples and edge cases pass.")
````


4. Walk through your Python code in plain words and explain why your solution works.


**Idea.** A Python `list` (the List ADT: an ordered sequence with access by index) lets me read or overwrite any position in O(1) time. So instead of building a new list, I rearrange the existing one using two indices, often called *two pointers*:

- `read` scans the list once from left to right and looks at every value.
- `write` marks the position where the next non-zero value belongs. It only moves forward after a non-zero value has been placed.

**Walkthrough of the code**

1. Set `write = 0`, because the first non-zero value I find belongs at the front of the list.
2. Loop `read` over every index `0, 1, ..., n - 1`.
3. If `nums[read]` is `0`, do nothing. `write` stays where it is, so this zero is left behind for a later non-zero value to swap with.
4. If `nums[read]` is not `0`, swap `nums[write]` and `nums[read]`. The non-zero value moves forward into position `write`, and the value that was there (a zero) moves back to position `read`. Then add 1 to `write`.
5. When the loop ends, return `nums`. It has been rearranged in place, and returning it matches the starter code's `-> List[int]`.

**Trace of Example 1**, `nums = [0, 1, 0, 3, 12]`:

| `read` | `nums[read]` | Action | List after this step | `write` after |
|---|---|---|---|---|
| 0 | 0 | zero, skip | `[0, 1, 0, 3, 12]` | 0 |
| 1 | 1 | swap positions 0 and 1 | `[1, 0, 0, 3, 12]` | 1 |
| 2 | 0 | zero, skip | `[1, 0, 0, 3, 12]` | 1 |
| 3 | 3 | swap positions 1 and 3 | `[1, 3, 0, 0, 12]` | 2 |
| 4 | 12 | swap positions 2 and 4 | `[1, 3, 12, 0, 0]` | 3 |

Result: `[1, 3, 12, 0, 0]`, which matches the expected output.

**Why it works.** At the start of every loop step, the list is split into three parts:

1. `nums[:write]` holds the non-zero values found so far, in the order `read` found them.
2. `nums[write:read]` holds only zeros: the zeros `read` has already passed.
3. `nums[read:]` has not been looked at yet.

Each step keeps this true. A zero simply becomes part of the zero block. A non-zero value is swapped to the end of the first part, and the zero it trades places with goes to the end of the zero block. (When no zero has been seen yet, `write == read` and the value swaps with itself, which changes nothing.)

- **Order is kept:** non-zero values are placed into positions `0, 1, 2, ...` in exactly the order they appear. Once placed, they are never moved again, because every later swap uses positions at or after `write`.
- **Nothing is lost or duplicated:** a swap only exchanges two values, so the list always has the same values and the same number of zeros.
- **Zeros end up last:** when `read` reaches the end, the unseen part is empty, so the list is all the non-zero values in their original order followed by all the zeros.


5. Explain the time and space complexity of your solution.


Let **n** be the length of `nums`.

**Time complexity: O(n)**
- There is one `for` loop and no nested loop. It runs exactly n times, once per index (Lecture 1: one loop over the input gives O(n)).
- Each step does a constant amount of work: one comparison with `0` and at most one swap and one addition. Reading or assigning `nums[i]` by index is O(1) for a Python list (Lecture 2), so each swap is O(1).
- In total that is n steps × O(1) work = **O(n)**.
- **Best, average and worst case are all O(n).** The loop always visits every element, whether the list has no zeros, all zeros, or a mix. This is also the best any correct solution can do: every element has to be checked at least once, because any value we skip could be a zero.
- For comparison, a common first idea is to find each zero, remove it with `pop(i)`, then `append(0)`. That is **O(n²)** in the worst case. Removing from the middle of a list shifts every later element one place left, which is O(n) per removal (Lecture 2), and there can be up to n zeros.

**Space complexity: O(1) extra space**
- The only extra memory is two integer variables, `read` and `write`. Their size doesn't depend on n.
- No new list is created. Values are swapped inside the original list (in place), the same way insertion sort uses O(1) space. Merge sort, by contrast, needs O(n) space for its temporary lists (Lecture 3).
- Counting the input list itself, total memory is O(n), as in the slides' example of summing a list of size n. That memory belongs to the input, though; the algorithm only adds O(1).


6. Describe an alternative solution to your problem in plain words, e.g., with a different data structure or time/space complexities. Note: you don't need to code your alternative solution, but a classmate reading this should be able to code it based on your description.


**Alternative: build a new list.** O(n) time, O(n) extra space.

Instead of swapping values inside the original list, copy them into a second list in the right order. A Python list used this way behaves like a queue: values are added at the back and keep the order they arrived in (first in, first out).

Steps a classmate could code directly:

1. Create an empty list `result` and a counter `zero_count = 0`.
2. Loop once over every value in `nums`.
    - If the value is not `0`, `append` it to `result`. This keeps the non-zero values in their original order.
    - If the value is `0`, add 1 to `zero_count` instead of storing it.
3. After the loop, append `0` to `result` `zero_count` times (for example, `result.extend([0] * zero_count)`).
4. Return `result`.

Example: for `[0, 1, 0, 3, 12]`, after step 2 `result = [1, 3, 12]` and `zero_count = 2`. Step 3 then gives `[1, 3, 12, 0, 0]`.

**Complexity**
- **Time: O(n).** There is one pass over the n values, and each `append` is O(1) (Lecture 2). Adding the zeros back is at most n more appends, so the total is O(n + n) = O(n).
- **Space: O(n) extra,** because `result` holds a full copy of the list.

**Trade-off versus my solution.** This version is arguably simpler to read, and it leaves the original list unchanged, which helps if the caller still needs it. But it uses O(n) extra memory, where my two-pointer solution uses O(1). Both take O(n) time, so the in-place two-pointer version is the better choice on space.
