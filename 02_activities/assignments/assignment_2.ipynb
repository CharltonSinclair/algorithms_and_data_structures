## Problem Three: Missing Number in Range
**Description**  
You are given a list containing `n` non-negative integers in the range `[1, n]`. Return a list of numbers that are missing from the range `[1, n]` in the given list. If there is no missing number, return -1. 

**Note**: integers in the list may not be unique.

**Examples**

* Example 1:

    Input: `lst = [1, 1, 3]`

    Output: [2]

* Example 2:

    Input: `lst = [5,3,2,4,1]`

    Output: -1

* Example 3:

    Input: `lst = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10]`

    Output: [4, 9]

**Problem 3 Starter Code**

````python
from typing import List
def missing_num(nums: List) -> int:
  # Code your solution here #
  """Return the numbers in [1, n] that are missing from the list, or -1 if nothing is missing."""
  n = len(nums)
  present = set(nums) 
  missing = [x for x in range(1, n + 1) if x not in present]  
  return missing if missing else -1


print(missing_num([1, 1, 3]))                        # expected [2]
print(missing_num([5, 3, 2, 4, 1]))                  # expected -1
print(missing_num([6, 8, 2, 3, 5, 7, 1, 1, 2, 10]))  # expected [4, 9]
````

Paramdeep's problem gives a string made only of the six bracket characters `(`, `)`, `[`, `]`, `{` and `}`, and asks whether the brackets are correctly matched. The function returns `True` or `False`.

A string is valid when both rules hold:
- **Same type:** every opening bracket is closed by its own type, so `(` only closes with `)`, `[` with `]` and `{` with `}`.
- **Right order:** brackets close in the correct order they opened. The most recently opened bracket must close first.


2. Create one new example with corresponding output to demonstrate that you have understood their assigned problem.


* Example: every bracket that closes matches, but one bracket is never closed
    ```python
    Input: s = "{[]()"
    Output: False
    ```

3. Walkthrough one example that your partner made and explain it.

Paramdeep's first example is `s = "([]{})"` → `True` (the same string as the notebook's provided Example 1). His solution reads the string left to right and keeps a stack of the brackets that are still open.

Every closing bracket matched the most recently opened bracket, and the stack is empty at the end, so the function returns `True`. The `[]` and `{}` pairs sit inside the outer `( )`, which is allowed.


4. Copy the solution your partner wrote. 


````python
def is_valid_brackets(s):
    pairs = {')': '(', '}': '{', ']': '['}
    stack = []
    for char in s:
        if char in pairs.values():
            stack.append(char)
        elif char in pairs.keys():
            if not stack or stack[-1] != pairs[char]:
                return False
            stack.pop()
    return not stack
````

5. Explain why their solution works in your own words.


His solution works because a stack has the same last-in, first-out (LIFO) order that the bracket rules require. The bracket that must close next is always the most recently opened one that is still open, and that is exactly the top of the stack.

- *The `pairs` dictionary maps each closing bracket to the opening bracket it needs. One lookup, `pairs[char]`, tells him what the top of the stack must be.

Those three checks cover every way a bracket string can be invalid, so if none of them fails, the string is valid. 

6. Explain the time and space complexity of their solution in your own words and identify if they found the correct time and space complexities.


Let n be the length of `s`.

Time: O(n).
- The `for` loop looks at each character once, and each step does a constant amount of work:
    - `char in pairs.values()` checks a fixed 3-item view, so it is constant time.
    - `char in pairs.keys()` and `pairs[char]` are O(1) dictionary lookups (Lecture 3).
    - `stack.append`, `stack[-1]` and `stack.pop()` all work at the end of the Python list, which is O(1). This is why a stack uses the back of a list rather than the front (Lecture 2).
- In total, n steps × O(1) = O(n).
- Worst case O(n): the whole string is scanned when it is valid, or when the problem only shows up at the end.
- Best case O(1): the function can stop as soon as it finds a mismatch, for example when the first character is a closing bracket.

Space: O(n).
- The stack can hold up to n items in the worst case, when the string is all opening brackets like `"(((((("`.
- The `pairs` dictionary has a fixed 3 entries, which is O(1).

He gives O(n) time, because each character is processed once with constant-time push and pop, and O(n) worst-case space, because the stack may hold every opening bracket. The only things he could add are that the early `return False` makes the best case faster, and that O(n) is the best possible time, since every character must be checked before a string can be called valid.
